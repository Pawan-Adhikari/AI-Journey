Cross Validation is basically splitting the total dataset in more reasonable way and testing the training performance on different splits we had made. If splits are random (80:20 train vs validation), certain splits would give better model performance whereas certain ones would be worse. To understand the true performance of the model, we split the data in a more predictable way. We first make divide the data into folds. For a 1000 rows dataset, dividing into 10 folds would mean 100 rows for each fold. Then we use each fold as a validation set in each unique run such that we train the model 10 times total. Finally, we compare or average the error metric from each run to understand the absolute performance of the model and data.

Test_data or data, either way, we now have processed data. Now, we need to separate the features(Independent variables) and target(Dependent variables) as X and Y. 

```python
X = data.drop('charges', axis = 1)
X.head()
```

```python
Y = data['charges']
print(Y)
```

```python
#X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.20, random_state=42)
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.20)
```

We will now have to create a regression model to train our data on.

```python
regr = linear_model.LinearRegression()
```

Then simply train the model on our dataset.

```python
scores = cross_val_score(
    regr,
    X,
    Y,
    cv = 10,
    scoring = "neg_mean_squared_error"
)
positive_mse_scores = -scores
rmse_scores = np.sqrt(positive_mse_scores)

regr.fit(X_train, Y_train)
```

Then simlpy run our validation/test set on it!

```python
Y_predicted = regr.predict(X_test)
```

Now, Plot the results and also show some key statistics.

```python
print(type(Y_test))
type(Y_predicted)
```

```python
plt.scatter(Y_test,Y_predicted)
plt.xlabel("Actual Charges ($)")
plt.ylabel("Predicted Charges ($)")
plt.title("Actual vs. Predicted Insurance Charges")
plt.plot([min(Y_test), max(Y_test)], [min(Y_test), max(Y_test)], color='red', linestyle='--')
plt.show()
print("Root mean squared error(This run): ", mean_squared_error(Y_predicted, Y_test) ** 0.5)
print("R2 score: (This run)", r2_score(Y_predicted, Y_test))
sum = 0.0
for i in rmse_scores:
    sum = sum + i
rmse = sum/10
print("Cross validation RMSE: ", rmse)
```
