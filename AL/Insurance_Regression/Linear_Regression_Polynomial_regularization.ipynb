Test_data or data, either way, we now have processed data. Now, we need to separate the features(Independent variables) and target(Dependent variables) as X and Y. 

```python
X = data.drop('charges', axis = 1)
X.head()
```

```python
Y = data['charges']
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.20, random_state=42)
```

```python
poly = PolynomialFeatures(degree=5, include_bias=False)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
s = StandardScaler()
X_train_poly_ss = s.fit_transform(X_train_poly)
X_test_poly_ss = s.transform(X_test_poly)
```

We will now have to create a regression model to train our data on.

```python
lasso = Lasso()
ridge = Ridge()
```

Then simply train the model on our dataset.

```python
pipeline_lasso = Pipeline([
                    ("poly", poly),
                    ("scaler", StandardScaler()),
                    ("model", lasso)
                    ])
pipeline_ridge = Pipeline([
                    ("poly", poly),
                    ("scaler", StandardScaler()),
                    ("model", ridge)
                    ])

scores_lasso = cross_validate(
    pipeline_lasso,
    X,
    Y,
    cv = 10,
    scoring = ["neg_mean_squared_error", "explained_variance"]
)
scores_ridge = cross_validate(
    pipeline_ridge,
    X,
    Y,
    cv = 10,
    scoring = ["neg_mean_squared_error", "explained_variance"]
)
positive_mse_scores_lasso = -scores_lasso['test_neg_mean_squared_error']
rmse_scores_lasso = np.sqrt(positive_mse_scores_lasso)

positive_mse_scores_ridge = -scores_ridge['test_neg_mean_squared_error']
rmse_scores_ridge = np.sqrt(positive_mse_scores_ridge)

lasso.fit(X_train_poly_ss, Y_train)
ridge.fit(X_train_poly_ss, Y_train)
```

Then simlpy run our validation/test set on it!

```python
Y_predicted_lasso = lasso.predict(X_test_poly_ss)
print("Coefficients lasso: ", lasso.coef_)
print("No. of coefficients lasso: ", len(lasso.coef_))
Y_predicted_ridge = ridge.predict(X_test_poly_ss)
print("Coefficients ridge: ", ridge.coef_)
print("No. of coefficients ridge: ", len(ridge.coef_))
```

Now, Plot the results and also show some key statistics.

```python
plt.scatter(Y_test,Y_predicted_lasso)
plt.xlabel("Actual Charges ($)")
plt.ylabel("Predicted Charges ($)")
plt.title("Actual vs. Predicted Insurance Charges")
plt.plot([min(Y_test), max(Y_test)], [min(Y_test), max(Y_test)], color='red', linestyle='--')
plt.show()
print("Root mean squared error(This run _lasso): ", mean_squared_error(Y_predicted_lasso, Y_test) ** 0.5)
print("R2 score: (This run _lasso)", r2_score(Y_predicted_lasso, Y_test))
sum = 0.0
for i in rmse_scores_lasso:
    sum = sum + i
rmse_lasso = sum/10
print("Cross validation RMSE, Polynomial Regression _lasso: ", rmse_lasso)

plt.scatter(Y_test,Y_predicted_ridge)
plt.xlabel("Actual Charges ($)")
plt.ylabel("Predicted Charges ($)")
plt.title("Actual vs. Predicted Insurance Charges")
plt.plot([min(Y_test), max(Y_test)], [min(Y_test), max(Y_test)], color='red', linestyle='--')
plt.show()
print("Root mean squared error(This run_ridge): ", mean_squared_error(Y_predicted_ridge, Y_test) ** 0.5)
print("R2 score: (This run_ridge)", r2_score(Y_predicted_ridge, Y_test))
sum = 0.0
for i in rmse_scores_ridge:
    sum = sum + i
    print(i)
rmse_ridge = sum/10
print("Cross validation RMSE, Polynomial Regression _ridge: ", rmse_ridge)
```
